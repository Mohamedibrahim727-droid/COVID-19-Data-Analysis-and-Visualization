```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os # Import os for path manipulation

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

extract_path = '/content/extracted_data' # Define the extraction path

df = pd.read_csv(os.path.join(extract_path, 'country_wise_latest.csv'))
df_day = pd.read_csv(os.path.join(extract_path, 'day_wise.csv'))

df_day['Date'] = pd.to_datetime(df_day['Date'])

df.drop_duplicates(inplace=True) # Changed df_country to df for consistency
num_cols = ['Confirmed', 'Deaths', 'Recovered', 'Active']
for col in num_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0) # Changed df_country to df for consistency
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(np.log10(df['Confirmed'] + 1), bins=20, kde=True, color='purple')
plt.title('Log-Transformed Distribution of Confirmed Cases')
plt.xlabel('Log10(Confirmed Cases + 1)')
plt.ylabel('Frequency')
plt.show()


top10_cases = df.nlargest(10, 'Confirmed')
plt.figure(figsize=(12, 6))
sns.barplot(data=top10_cases, x='Confirmed', y='Country/Region', palette='Reds_r')
plt.title('Top 10 Affected Countries by Confirmed Cases')
plt.xlabel('Confirmed Cases')
plt.ylabel('Country')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df_day['Date'], df_day['Confirmed'], label='Confirmed Cases', color='blue', linewidth=2)
plt.plot(df_day['Date'], df_day['Deaths'], label='Deaths', color='red', linewidth=2)
plt.plot(df_day['Date'], df_day['Recovered'], label='Recovered', color='green', linewidth=2)
plt.title('Global COVID-19 Trends Over Time')
plt.xlabel('Date')
plt.ylabel('Count')
plt.legend()
plt.show()

plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Confirmed', y='Deaths', hue='WHO Region', size='Active', sizes=(20, 200), alpha=0.8)
plt.title('Confirmed Cases vs. Deaths per Country')
plt.xlabel('Confirmed Cases')
plt.ylabel('Deaths')
plt.show()

plt.figure(figsize=(8, 6))
corr_matrix = df[num_cols].corr()
sns.heatmap(corr_matrix, annot=True, cmap='Blues', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix of COVID-19 Metrics')
plt.show()

df['Recovery Rate (%)'] = (df['Recovered'] / df['Confirmed']) * 100
top10_rec = df[df['Confirmed'] > 50000].nlargest(10, 'Recovery Rate (%)')
plt.figure(figsize=(12, 6))
sns.barplot(data=top10_rec, x='Recovery Rate (%)', y='Country/Region', palette='Greens_r')
plt.title('Top 10 High Recovery Rate Countries (Confirmed > 50k)')
plt.xlabel('Recovery Rate (%)')
plt.ylabel('Country')
plt.show()

region_df = df.groupby('WHO Region')[['Confirmed', 'Deaths', 'Recovered']].sum().reset_index()
region_df.plot(x='WHO Region', kind='bar', stacked=True, figsize=(12, 6), colormap='Set2')
plt.title('COVID-19 Totals by WHO Region')
plt.xlabel('WHO Region')
plt.ylabel('Total Count')
plt.xticks(rotation=45)
plt.show()

df['Death Rate (%)'] = (df['Deaths'] / df['Confirmed']) * 100
df['Impact_Score'] = (np.log10(df['Confirmed'] + 1) * 0.4) + (np.log10(df['Deaths'] + 1) * 0.4) - (df['Recovery Rate (%)'] * 0.02)

def classify_impact(score):
    if score >= 5.0:
        return 'High Impact'
    elif score >= 3.0:
        return 'Medium Impact'
    else:
        return 'Low Impact'

df['Impact Category'] = df['Impact_Score'].apply(classify_impact)

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Impact Category', order=['Low Impact', 'Medium Impact', 'High Impact'], palette='Oranges')
plt.title('Country Classification by COVID-19 Impact')
plt.xlabel('Impact Category')
plt.ylabel('Number of Countries')
plt.show()
```
